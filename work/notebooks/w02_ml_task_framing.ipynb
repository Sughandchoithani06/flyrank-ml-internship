## 3. Success metric

*One metric you can defend. What number means 'good'?*

queue, and a reviewer only has time to check a limited number of pages
— so what matters is how many of the top 50 recommended pages are
actually true positives, not how the model does overall. The starter
pipeline already shows this matters: the baseline rule gets 0.240
Precision@50 (about 12 of 50 correct), while a random forest gets 0.740
(about 37 of 50) — that gap is the whole reason this metric fits the
real decision better than accuracy would.

## 5. Why ML beats a fixed rule here

*What makes the pattern too messy for an if-statement?*
A fixed rule (like the starter baseline score) can only combine a
handful of hand-picked thresholds in a way a human decided in advance —
it can't learn how those signals actually interact with each other
across thousands of pages. The starter pipeline already proves this: the
hand-written baseline rule reaches Precision@50 of 0.240, but a random
forest trained on the same signals reaches 0.740 — roughly three times
better at picking the right pages to review first. That gap is real
evidence that there's a pattern in this data too complex for a fixed
rule to capture, which is exactly the kind of problem ML is for.
